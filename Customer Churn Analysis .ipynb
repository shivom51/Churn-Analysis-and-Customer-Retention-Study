```python
# groupby to get counts
group = df.groupby(['SeniorCitizen', 'Churn']).size().unstack()

# convert to percentage
percent = group.div(group.sum(axis=1), axis=0) * 100

plt.figure(figsize=(6,4))

# bottom part (Churn = No)
plt.bar(percent.index, percent['No'], label="Churn = No")

# top part (Churn = Yes)
plt.bar(percent.index, percent['Yes'], bottom=percent['No'], label="Churn = Yes")

# add percentage labels
for i, idx in enumerate(percent.index):
    # label for 'No'
    plt.text(i, percent['No'][idx]/2, 
             f"{percent['No'][idx]:.1f}%", 
             ha='center', va='center', color='white', fontsize=10)
    
    # label for 'Yes'
    plt.text(i, percent['No'][idx] + percent['Yes'][idx]/2,
             f"{percent['Yes'][idx]:.1f}%", 
             ha='center', va='center', color='white', fontsize=10)

plt.ylabel("Percentage")
plt.title("SeniorCitizen vs Churn (Stacked %)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# columns you want to plot
cols = [
    "PhoneService", "MultipleLines", "InternetService",
    "OnlineSecurity", "OnlineBackup", "DeviceProtection",
    "TechSupport", "StreamingTV", "StreamingMovies"
]

plt.figure(figsize=(15,12))        # big figure for 9 plots

# loop through columns with index
for i, col in enumerate(cols, 1):
    
    plt.subplot(3, 3, i)           # create 3 rows, 3 columns
    sns.countplot(x=col, data=df,hue="Churn")  # countplot

    plt.title(col)                 # title
    plt.xlabel("")                 # hide x-label (clean look)
    plt.ylabel("")                 # hide y-label
    #plt.xticks(rotation=20)        # angle for readability

plt.tight_layout()                 # adjust spacing
plt.show()                         # show all 9 together
```

#Most customers who did not churn appear in the “No” category for almost every service feature like OnlineSecurity, OnlineBackup, TechSupport, and DeviceProtection.
#Churn is noticeably higher among customers who lack these services (especially “No internet service” and “No protection” categories).
#Customers who have internet-related services show more churn compared to phone-based services.
#Overall, missing security and support services strongly relate to higher churn.

```python
plt.figure(figsize=(9,4))
ax=sns.countplot(x='PaymentMethod',data=df,hue="Churn")
ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[1])
plt.show()
```

#2. Customers paying through **Mailed Check, Bank Transfer, and Credit Card (automatic)** churn far less, suggesting **automatic or stable #payment methods** are linked to better customer retention.
